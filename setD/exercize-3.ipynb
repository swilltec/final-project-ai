Consider the following scenario. You live in an area where earthquakes happen on the average every 111th day. Another annoying thing is that a burglar breaks into your home on the average once a year. Both events occur independently of each other and uniformly throughout the year (so any day is like another).

In case of an earthquake, your home alarm goes off with probability 0.81. If your home is broken into, the alarm goes off with probability 0.92. If these both should occur at the same time, the alarm goes off with probability 0.97. If nothing special is going on, there's a false alarm with probability 0.0095.

Now modify the algorithm from the previous exercise so that you can generate data from variables [E]arthquake, [B]urglar, [A]larm. Generate a sample of n=100 000 tuples.

Among the tuples with A=1, what is the fraction where B=1?
Among the tuples with A=1 and E=1, what is the fraction where B=1?
and (don't forget these):

Give an interpretation to your answers for items 1 & 2. What probabilities do they approximate?
Are they in line with your intuition? In other words, do they make sense? In particular: which of the fractions is bigger and why?
Repeat the experiment a couple of times to get a feeling of how much the results change just by chance. (This is the nature of the Monte Carlo approximation.) Experiment with different values of n. How does it affect the variability?

### Variables

- \(E\): Earthquake occurs
- \(B\): Burglary occurs
- \(A\): Alarm goes off

Each variable is binary:
- \(1\): Event occurs
- \(0\): Event does not occur

### Probabilities

$$
\begin{aligned}
P(E) &= \frac{1}{111} \\
P(B) &= \frac{1}{365} \\
P(A \mid E \land \neg B) &= 0.81 \\
P(A \mid B \land \neg E) &= 0.92 \\
P(A \mid E \land B) &= 0.97 \\
P(A \mid \neg E \land \neg B) &= 0.0095
\end{aligned}
$$

```python
# independent events
p_earthquake = 1/111
p_burglary = 1/365

# conditionals
p_alarm_earthquake_only = 0.81
p_alarm_burglary_only = 0.92
p_alarm_both = 0.97
p_alarm_neither = 0.0095
```

```python
import random
```

```python
n = 100_000
sample = list()

def binary_roll(probability):
    return 1 if random.random() < probability else 0

# Fill sample n times
for i in range(n):
    earthquake = binary_roll(p_earthquake)
    burglary = binary_roll(p_burglary)
    alarm = 0

    if earthquake and not burglary:
        alarm = binary_roll(p_alarm_earthquake_only)
    elif not earthquake and burglary:
        alarm = binary_roll(p_alarm_burglary_only)
    elif earthquake and burglary:
        alarm = binary_roll(p_alarm_both)
    else:
        alarm = binary_roll(p_alarm_neither)

    sample.append((earthquake, burglary, alarm))

print(f'Sample filled with {n} entries.')
```

### Sample Structure

The `sample` list contains **100,000 tuples**, each representing one simulated day.

Each tuple has the format `(E, B, A)`, where:

- `E`: Earthquake
- `B`: Burglary
- `A`: Alarm

Each variable is binary: `1` means the event occurred, and `0` means it did not.

For example, `(1, 0, 1)` means an earthquake occurred, no burglary occurred, and the alarm went off.

```python
# Count all samples where A, A AND B
sample_alarm_count = 0
sample_alarm_and_burglary_count = 0
for tup in sample:
    if tup[2]: # A
        sample_alarm_count += 1

    if tup[2] and tup[1]: # A AND B
        sample_alarm_and_burglary_count += 1

print('Among the tuples with A=1, what is the fraction where B=1?')

# Avoid divide by 0
if sample_alarm_count > 0:
    answer_1 = sample_alarm_and_burglary_count / sample_alarm_count
    print(answer_1)
else:
    print("No alarm events occurred. Cannot estimate probability.")
```

```python
# Count all samples where A AND E, A AND E AND B 
sample_alarm_and_earthquake_count = 0
sample_all_count = 0

for tup in sample:
    if tup[2] and tup[0]: # A AND E
        sample_alarm_and_earthquake_count += 1

    if tup[0] and tup[1] and tup[2]: # A AND E AND B
        sample_all_count += 1

print('Among the tuples with A=1 and E=1, what is the fraction where B=1?')

# avoid divide by 0
if sample_alarm_and_earthquake_count > 0:
    answer_2 = sample_all_count / sample_alarm_and_earthquake_count
    print(answer_2)
else:
    print("No alarm and earthquake events occurred. Cannot estimate probability.")
```

### Interpretation of Results

**1. What probabilities do the results approximate?**

- **Answer 1:** \(P(B=1 | A=1)\), the probability of a burglary given that the alarm went off.
- **Answer 2:** \(P(B=1 | A=1, E=1)\), the probability of a burglary given that both the alarm went off and an earthquake occurred.

**2. Which fraction is larger, and why?**

The first fraction is larger. If the alarm goes off without knowing whether an earthquake occurred, a burglary is a possible explanation. However, if we know an earthquake occurred, there is already another likely reason for the alarm, making a burglary less probable.

**3. How does sample size affect variability?**

Monte Carlo estimates change slightly between runs because the events are randomly generated. With smaller sample sizes, the results tend to fluctuate more. As the sample size increases, the estimates generally become more stable and approach the true probabilities. This effect is especially noticeable for rare events.
